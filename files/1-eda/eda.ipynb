# Hipóteses

| Hipótese | Explicação |
| -- | -- |
| H1: de passagem e desengajados| Entre os clientes em churn, há clientes de passagem, que compraram uma única vez, e há clientes que tiveram mais de uma compra e se desengajaram.|
| H2.1: frequência | O intervalo de compra (inter-purchase time) entre clientes desengajados aumenta até que efetivamente entrem em churn|

# H1: entre os clientes em churn num dado momento, há uma parcela de passagem e há clientes verdadeiramente desengejados

```python
# calcular quantos clientes em churn no ultimo dia do ano, quantos desses são one-timers e quanto sao recorrentes
base['DATA_ATEND'] = pd.to_datetime(base['DATA_ATEND'])
max_date = base['DATA_ATEND'].max()

threshold = max_date - pd.Timedelta(days=90)

threshold
```

```python
resumo_clientes = base.groupby('CLI_CPF').agg(visitas = ('DATA_ATEND', 'nunique'), ultima_compra= ('DATA_ATEND', 'max'))
resumo_clientes
```

```python
resumo_clientes['churn'] = resumo_clientes['ultima_compra'] < threshold
resumo_clientes['tipo'] = resumo_clientes['visitas'].apply(lambda x: 'one timer' if x == 1 else 'recurring')
resumo_clientes
```

```python
churn_df = resumo_clientes[resumo_clientes['churn']]
print(churn_df['tipo'].value_counts())
print(churn_df['tipo'].value_counts(normalize=True) * 100)
```

# H2.1 - O intervalo entre as compras aumenta antes do churn

Isto é, um cliente se desengaja gradualmente

```python
# pegar as as 3 últimas visitas
base['DATA_ATEND'] = pd.to_datetime(base['DATA_ATEND'])
visitas = base[['CLI_CPF', 'DATA_ATEND']].drop_duplicates().sort_values(['CLI_CPF', 'DATA_ATEND'])

visitas['intervalo'] = visitas.groupby(by=['CLI_CPF'])['DATA_ATEND'].diff().dt.days

visitas
```

```python
# ordena as datas das últimas compras pra um cliente de forma decrescente. ex cliente 000068324b7a, última compra é 0, penúltima é 1, 
# ...
# ideia: pegar o intervalo de 0 e 1, 1 e 2, 2 e 3, 3 e 4
visitas['ordem_rev'] = visitas.groupby('CLI_CPF').cumcount(ascending=False)
visitas.head(10)
```

```python
ultimas_compras = visitas[visitas['ordem_rev'] <= 2].pivot(
    index='CLI_CPF',
    columns = 'ordem_rev',
    values='intervalo'
)

ultimas_compras.columns = ['ultimo_intervalo', 'penultimo_intervalo', 'antepenultimo_intervalo']

data_corte = base['DATA_ATEND'].max() - pd.Timedelta(days=90)
ultimas_datas = visitas.groupby('CLI_CPF')['DATA_ATEND'].max()
ultimas_compras['churn'] = ultimas_datas < data_corte

ultimas_compras
```

```python

df_h21 = ultimas_compras.dropna()

# 7. Comparar médias e medianas dos intervalos entre Ativos (False) e Churn (True)
print("--- Médias dos Intervalos (em dias) ---")
print(df_h21.groupby('churn')[['antepenultimo_intervalo', 'penultimo_intervalo', 'ultimo_intervalo']].mean().round(1))

print("\n--- Medianas dos Intervalos (em dias) ---")
print(df_h21.groupby('churn')[['antepenultimo_intervalo', 'penultimo_intervalo', 'ultimo_intervalo']].median())
```

Resultado não discriminatório entre clientes em churn ou não

# H2.3 Meses com mais feriados apresentam maior taxa de recorrência

```python

base['DATA_ATEND'] = pd.to_datetime(base['DATA_ATEND'])
base['MES'] = base['DATA_ATEND'].dt.month

primeira_compra = base.groupby('CLI_CPF')['DATA_ATEND'].min()
base['primeira_compra'] = base['CLI_CPF'].map(primeira_compra)

# Cliente de retorno: qualquer compra realizada após a sua primeira data histórica
base['is_retorno'] = base['DATA_ATEND'] > base['primeira_compra']

# 3. Métricas mensais
mensal = base.groupby('MES').agg(
    faturamento=('FATUR_VENDA', 'sum'),
    clientes_totais=('CLI_CPF', 'nunique'),
    clientes_retorno=('CLI_CPF', lambda x: x[base.loc[x.index, 'is_retorno']].nunique())
)

# 4. Inserir a contagem de feriados de 2024
feriados_2024 = {
    1: 1, 2: 2, 3: 1, 4: 1, 5: 2, 6: 0,
    7: 0, 8: 0, 9: 1, 10: 1, 11: 3, 12: 1
}
mensal['qtd_feriados'] = mensal.index.map(feriados_2024)

# 5. Visualizar tabela e correlação
print(mensal)

print("\nCorrelação entre Feriados e Clientes de Retorno:")
print(mensal['qtd_feriados'].corr(mensal['clientes_retorno']).round(3))
```

```python
print("\nCorrelação entre Feriados e Clientes de Retorno:")
print(mensal['qtd_feriados'].corr(mensal['faturamento']).round(3))
```

O faturamento é impactado positivamente pela quantidade de feriados, mas isso é movido pela Páscoa e Natal principalmente. Não há evidência forte o suficiente para explicar que mais clientes voltem na loja num feriado.

# H3 tipo de filial

```python

perfil_filial = base.groupby('CLI_CPF').agg(
    total_visitas=('DATA_ATEND', 'nunique'),
    ultima_compra=('DATA_ATEND', 'max'),
    filiais=('FILIAL', 'unique'),
    faturamento_total=('FATUR_VENDA', 'sum')
)

perfil_filial['churn'] = perfil_filial['ultima_compra'] < threshold
perfil_filial['one_timer'] = perfil_filial['total_visitas'] == 1

# 3. Classificar o canal de preferência do cliente
def definir_canal(filiais):
    if len(filiais) > 1:
        return 'Ambos (Híbrido)'
    return filiais[0]

perfil_filial['tipo_filial'] = perfil_filial['filiais'].apply(definir_canal)

# 4. Tabela comparativa por Tipo de Filial
comparativo_h3 = perfil_filial.groupby('tipo_filial').agg(
    total_clientes=('total_visitas', 'count'),
    taxa_churn_pct=('churn', lambda x: (x.mean() * 100)),
    taxa_one_timer_pct=('one_timer', lambda x: (x.mean() * 100)),
    media_visitas_ano=('total_visitas', 'mean'),
    gasto_medio_anual=('faturamento_total', 'mean')
).round(2)

print(comparativo_h3)
```

shoppings têm mais clientes one timers. clientes de rua visitam mais. clientes multicanais têm menos taxa de churn e mais visitas por ano
